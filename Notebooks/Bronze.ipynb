# **Incremental Data Ingestion**

```python
dbutils.widgets.text("src_folder", "")
schemaname = dbutils.widgets.get("src_folder")
```

```python
df = spark.readStream.format("cloudFiles") \
                     .option("cloudFiles.format", "csv") \
                     .option("cloudFiles.schemaLocation", f"/Volumes/databricksete/bronze/bronzevolume/{schemaname}/checkpoint") \
                     .option("cloudFiles.schemaEvolutionMode", "rescue") \
                     .load(f"/Volumes/databricksete/raw/rawvolume/rawdata/{schemaname}/")
```

```python
df.writeStream.format("delta") \
              .outputMode("append") \
              .trigger(once=True) \
              .option("checkpointLocation", f"/Volumes/databricksete/bronze/bronzevolume/{schemaname}/checkpoint") \
              .option("path", f"/Volumes/databricksete/bronze/bronzevolume/{schemaname}/data") \
              .start()
```
